```python
#Count the occurences of each word in a line of text.
x = input("Enter a line of text: ")
words = x.split()
count = {}
for word in words:
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1
for word in count:
    print(word, ":", count[word])
```

```python
#Prompt the user for a list of integers .
#For all values greater than 100,store 'over' instead
numbers = input("Enter numbers separated by spaces: ").split()
for i in range(len(numbers)):
    numbers[i] = int(numbers[i])
for i in range(len(numbers)):
    if numbers[i] > 100:
        numbers[i] = "over"
print(numbers)
```

```python
#Store a list of first names. Count the occurrences 'a' within a list
names = input("Enter names separated by spaces: ").split()
print("List:",names)
count = 0
for name in names:
    for letter in name:
        if letter == "a" or letter == "A":
            count = count + 1
print("Number of a's:", count)
```

```python
#Find gcd of 2 numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
gcd = 1
for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i
print("GCD:", gcd)
```

```python
#From a list of integers, create a list removing even numbers
numbers = list(map(int,input("Enter numbers separated by spaces: ").split()))
new_list = []
for num in numbers:
    if num % 2 != 0:
        new_list.append(num)
print("List after removing even numbers:", new_list)
```

```python
#Program to find factorial of a number
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n + 1):
    fact = fact * i
print("Factorial =", fact)
```

```python
#Program to find factorial of a number using while
n = int(input("Enter a number: "))
fact = 1
i=1
while i <= n:
    fact=fact*i
    i=i+1
print("Factorial =", fact)
```

```python
#Generate Fibanocci series of N terms
n = int(input("Enter the number of terms: "))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
#Generate a list of four digit numbers in a given range with all their digits even and the number is a perfect square
import math
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
for n in range(start, end + 1):
    if 1000 <= n <= 9999:
        root = math.isqrt(n)
        if root * root == n:
            temp = n
            all_even = True
            while temp > 0:
                digit = temp % 10
                if digit % 2 != 0:
                    all_even = False
                    break
                temp = temp // 10
            if all_even:
                print(n)
```
