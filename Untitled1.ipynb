```python
#Write a pyton program which accepts a sequence of comma-seperated numbers from user
#Generate a tuple with those numbers 
#Find average of all items in that tuple
num=tuple(map(int,input("Enter a list of numbers seperated by comma :").split(",")))
print("Tuple :",num)
average= sum(num)/len(num)
print("Average of numbers in the tuple :",average)
```

```python
set1 = set(input("Enter colors for Set 1 (separated by spaces): ").split())
set2 = set(input("Enter colors for Set 2 (separated by spaces): ").split())

# Colors in Set 1 but not in Set 2
difference = set1 - set2

# Colors in both sets
intersection = set1 & set2

# All colors from both sets
union = set1 | set2

print("Set 1:", set1)
print("Set 2:", set2)
print("Set 1 - Set 2:", difference)
print("Colors in both:", intersection)
print("All colors:", union)
```

```python
students = {}

roll1 = int(input("Enter roll number 1: "))
name1 = input("Enter name 1: ")
students[roll1] = name1

roll2 = int(input("Enter roll number 2: "))
name2 = input("Enter name 2: ")
students[roll2] = name2

roll3 = int(input("Enter roll number 3: "))
name3 = input("Enter name 3: ")
students[roll3] = name3

roll4 = int(input("Enter roll number 4: "))
name4 = input("Enter name 4: ")
students[roll4] = name4

roll5 = int(input("Enter roll number 5: "))
name5 = input("Enter name 5: ")
students[roll5] = name5

print("Original dictionary:", students)

# Ascending order
ascending = dict(sorted(students.items()))
print("Ascending order:", ascending)

# Descending order
descending = dict(sorted(students.items(), reverse=True))
print("Descending order:", descending)
```
