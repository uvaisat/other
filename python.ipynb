```python
#Count the occurences of each word in a line of text.
a=input("Enter a line of text:")
word= a.split()
count={}
for x in word:
    if x in count:
        count[x]=count[x]+1
    else:
        count[x]=1
print(count)
for x in count:
    print(x,":",count[x])
```

```python
#Store a list of first names. Count the occurrences 'a' within a list
w= input("Enter names")
count=0
for x in w:
        if x == 'a' or x =='A':
            count=count+1
print("Number of a's:", count)
```

```python
#Prompt the user for a list of integers .
#For all values greater than 100,store 'over' instead
num=list(map(int,input("Enter a list of numbers:").split()))
for x in range(len(num)):
    if num[x] > 100:
        num[x]="over"
print(num)
        
```

```python
#Find gcd of 2 numbers
a=int(input("Enter the  first number"))
b=int(input("Enter the  second number"))
gcd=1
for i in range(1,min(a,b)+1):
    if a % i==0 and b%i==0:
        gcd = i
print(gcd)
```

```python
#From a list of integers, create a list removing even numbers
old=list(map(int,input("Enter numbers:").split()))
new=[]
for x in old:
    if x% 2==1:
        new.append(x)
print(new)
```

```python
#Program to find factorial of a number
n=int(input("Number:"))
fact =1
for i in range(1,n+1):
    fact=fact*i
print("Factorial of ",n,"=",fact)
```

```python
#Generate Fibanocci series of N terms using while
n = int(input("Enter the number of terms: "))
a = 0
b = 1
i = 0
while i < n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    i = i + 1
```

```python
#Count the number of characters(character frequency) in a string.
string=input("Enter a string:")
d={}
for i in string:
    if i in d:
        d[i]=d[i]+1
    else:
        d[i]=1
print("Dict:",d)
```

```python
#Enter 2 lists of integers.
#Check(a)whether list are of same length(b) whether list sums to same value(c) whether any value occur in both.
list1=list(map(int,input("Enter list 1:").split()))
list2=list(map(int,input("Enter list 2:").split()))
if len(list1) == len(list2):
    print("Same length")
else:
    print("Not same length")
if sum(list1) == sum(list2):
    print("Same sum")
else:
    print("Not same sum")
common=False
for n in list1:
    if n in list2:
        common=True
        break
if common:
    print("Common elements occured")
else:
    print("No Common elements occured")
```
