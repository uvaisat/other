```python
#Count the occurences of each word in a line of text.
text= input("Enter a line of text:")
words= text.split()
count={}
for word in words:
    if word in count:
        count[word]=count[word]+1
    else:
        count[word]=1
print("Occurence")
for word in count:
    print(word,":",count[word])
```
